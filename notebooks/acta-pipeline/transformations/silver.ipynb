```sql
CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_usuario AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_usuario
WHERE id_dim_usuario IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_empresa AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_empresa
WHERE id_dim_empresa IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_data AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_data
WHERE id_dim_data IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_ciclo AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_ciclo
WHERE id_dim_ciclo IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_plano_acao AS
SELECT DISTINCT
    id_dim_plano_acao,
    id_dim_ciclo,
    nome_plano_acao,
    status_plano_acao,
    prioridade_plano_acao
FROM workspace.bronze.bronze_plano_acao
WHERE id_dim_plano_acao IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_fato_tarefas AS
SELECT
    id_dim_empresa,
    id_dim_usuario,
    id_dim_data,
    id_tarefa,
    COALESCE(qtd_tarefas, 0) AS qtd_tarefas,
    COALESCE(qtd_concluidas, 0) AS qtd_concluidas,
    COALESCE(qtd_em_andamento, 0) AS qtd_em_andamento,
    COALESCE(qtd_bloqueadas, 0) AS qtd_bloqueadas,
    COALESCE(qtd_atrasadas, 0) AS qtd_atrasadas
FROM workspace.bronze.bronze_fato_tarefas
WHERE id_tarefa IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_fato_plano AS
SELECT *
FROM workspace.bronze.bronze_fato_plano
WHERE id_plano_acao IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_fato_treinamentos AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_fato_treinamentos
WHERE id_treinamento IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_fato_alertas AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_fato_alertas
WHERE id_alerta IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_fato_logs AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_fato_logs
WHERE id_dim_usuario IS NOT NULL
  AND id_dim_data IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_dau_empresa AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_dau_empresa
WHERE id_empresa IS NOT NULL
  AND data_atv IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_dau_ciclos AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_dau_ciclos
WHERE id_empresa IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_tarefas_bloqueadas AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_tarefas_bloqueadas
WHERE id_tarefa IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_resumo_ciclo AS
SELECT DISTINCT
    id_ciclo,
    id_empresa,
    empresa AS nome_empresa,
    titulo,
    status,
    data_inicio,
    data_estimada_fim,
    data_fim_real,
    qtd_problemas,
    qtd_causas,
    qtd_planos,
    planos_concluidos,
    qtd_tarefas,
    tarefas_concluidas,
    tarefas_atrasadas,
    percentual_tarefas_concluidas,
    qtd_metas,
    metas_atingidas
FROM workspace.bronze.bronze_resumo_ciclo
WHERE id_ciclo IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_problema AS
SELECT id,
    TRIM(titulo) AS titulo,
    descricao
FROM workspace.bronze.bronze_problema
WHERE id IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_atv_usuario AS
SELECT DISTINCT *
FROM workspace.bronze.bronze_atv_usuario
WHERE id IS NOT NULL;

CREATE OR REFRESH MATERIALIZED VIEW workspace.silver.silver_endereco AS
SELECT id,
    id_empresa,
    cep,
    UPPER(uf) AS uf,
    TRIM(cidade) AS cidade,
    principal
FROM workspace.bronze.bronze_endereco_empresa
WHERE id IS NOT NULL AND uf IS NOT NULL;
```
